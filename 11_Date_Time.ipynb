# Notebook 11 – Date & Time Operations
## Learning objectives
Learn how to work with dates and times in Pandas: `datetime`, timestamps, date parsing, time differences, time zones, and extracting useful date features.

**Setup:** Run the import cell first. Examples use small, fictional datasets.

```python
import pandas as pd
from datetime import datetime, date, timedelta

print("Pandas version:", pd.__version__)
```

## 1. Datetime
**Definition:** Datetime values represent a specific calendar date and clock time. Pandas commonly stores them in the `datetime64[ns]` dtype.

**Real-world example:** Recording when a customer placed an order.
**Business example:** Measuring how long orders take to ship.
**AI/ML example:** Date/time values can become features for forecasting demand.

The example below creates a column of dates and checks its data type.

```python
orders = pd.DataFrame({
    "order_id": [101, 102, 103],
    "order_time": pd.to_datetime(["2026-10-10 09:30", "2026-10-11 14:15", "2026-10-12 18:45"])
})
print(orders)
print("\\nColumn type:", orders["order_time"].dtype)
```

## 2. Timestamp
**Definition:** A timestamp identifies a precise point in time. `pd.Timestamp` is Pandas' scalar type for a date and time.

**Real-world:** A timestamp marks when a sensor reading was recorded.
**Business:** Compare the exact time a payment was received.
**AI/ML:** Align event records before creating time-series features.

```python
one_timestamp = pd.Timestamp("2026-10-15 10:45:00")
print(one_timestamp)
print("Year:", one_timestamp.year)
print("Month:", one_timestamp.month)
print("Day:", one_timestamp.day)
```

## 3. Date Parsing
**Definition:** Date parsing converts text or other values into a recognized date/time type. Use `pd.to_datetime()`.

**Real-world:** Convert dates imported from a CSV.
**Business:** Sort invoices chronologically.
**AI/ML:** Parse date columns before time-series analysis.

`errors="coerce"` converts invalid dates to `NaT` (Pandas' missing datetime value).

```python
raw_dates = pd.Series(["2026-01-05", "15/02/2026", "not-a-date"])
parsed_dates = pd.to_datetime(raw_dates, format="mixed", dayfirst=True, errors="coerce")
print(parsed_dates)
```

## 4. Timedelta
**Definition:** A timedelta represents a duration or difference between two dates/times.

**Real-world:** Find the time between departure and arrival.
**Business:** Calculate delivery duration.
**AI/ML:** Use elapsed time since a user's last activity as a feature.

```python
shipments = pd.DataFrame({
    "ordered_at": pd.to_datetime(["2026-10-01 09:00", "2026-10-02 11:30"]),
    "delivered_at": pd.to_datetime(["2026-10-03 15:00", "2026-10-04 10:30"])
})
shipments["delivery_time"] = shipments["delivered_at"] - shipments["ordered_at"]
shipments["delivery_hours"] = shipments["delivery_time"].dt.total_seconds() / 3600
print(shipments)
```

## 5. Time Zones
**Definition:** A time zone describes the local clock convention for a region. Time-zone-aware timestamps help compare events across regions.

**Real-world:** A video call is scheduled across countries.
**Business:** Convert sales events into a consistent reporting zone.
**AI/ML:** Align timestamped logs from servers in different regions.

`tz_localize()` assigns a time zone to a naive timestamp; `tz_convert()` changes an already-aware timestamp to another zone.

```python
local_times = pd.Series(pd.to_datetime(["2026-10-15 09:00", "2026-10-15 17:00"]))
india_times = local_times.dt.tz_localize("Asia/Kolkata")
utc_times = india_times.dt.tz_convert("UTC")
print("India time:")
print(india_times)
print("\\nUTC time:")
print(utc_times)
```

## 6. Extracting Date Features
**Definition:** Date feature extraction creates useful columns such as year, month, day, weekday, hour, or quarter from a datetime column.

**Real-world:** Identify which weekdays have the most activity.
**Business:** Compare sales by month or quarter.
**AI/ML:** Build calendar features for demand or traffic forecasting.

Pandas' `.dt` accessor provides datetime components.

```python
sales = pd.DataFrame({
    "date": pd.to_datetime(["2026-01-05 09:20", "2026-02-14 16:10", "2026-02-16 12:00"]),
    "sales": [1200, 1800, 950]
})
sales["year"] = sales["date"].dt.year
sales["month"] = sales["date"].dt.month
sales["day"] = sales["date"].dt.day
sales["weekday"] = sales["date"].dt.day_name()
sales["hour"] = sales["date"].dt.hour
sales["quarter"] = sales["date"].dt.quarter
print(sales)
```
