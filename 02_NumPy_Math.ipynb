## 1. Sum — `np.sum()`

**Definition:** Adds all values in an array, or values along a selected axis.

**Simple example:** Add daily sales values.

**Real-world analogy:** Adding the money collected by a shop each day gives total sales.

**Business example:** A shop sold 10, 20, and 30 items on three days. The total is 60 items.

**AI/ML use case:** Summing losses across examples can help calculate a model's total error.

The code below demonstrates the concept because `np.sum()` adds every value in the array.

```python
daily_sales = np.array([10, 20, 30])
print("Daily sales:", daily_sales)
print("Total sales:", np.sum(daily_sales))
```

## 2. Mean — `np.mean()`

**Definition:** The arithmetic mean is the sum of values divided by the number of values.

**Formula:** Mean = sum of values / number of values.

**Analogy:** Sharing a total amount equally among people.

**Business example:** Average daily orders help a manager understand typical demand.

**AI/ML use case:** Means are used to summarize features and are part of some normalization methods.

Here, NumPy adds the three values and divides by three.

```python
daily_orders = np.array([12, 18, 15])
print("Average daily orders:", np.mean(daily_orders))
```

## 3. Median — `np.median()`

**Definition:** The median is the middle value after sorting the data. For an even number of values, it is the average of the two middle values.

**Analogy:** Line people up from shortest to tallest; the person in the middle represents the median position.

**Business example:** Median salary can describe a typical salary without being pulled strongly by a few very high salaries.

**AI/ML use case:** Median values can fill missing numerical data because they are less sensitive to extreme values than the mean.

The example includes an unusually large value to show why the median can differ from the mean.

```python
delivery_times = np.array([20, 22, 24, 25, 100])
print("Mean delivery time:", np.mean(delivery_times))
print("Median delivery time:", np.median(delivery_times))
```

## 4. Variance — `np.var()`

**Definition:** Variance measures how spread out values are from their mean by averaging squared differences from the mean.

**Population formula:** variance = average of (value − mean)².

**Analogy:** If students' marks are very different from the class average, the variance is larger.

**Business example:** A retailer can compare variation in daily demand across products.

**AI/ML use case:** Variance can help identify features that barely change and is used in statistical models.

NumPy uses population variance by default (`ddof=0`). Use `ddof=1` for sample variance.

```python
marks = np.array([70, 80, 90])
print("Population variance:", np.var(marks))
print("Sample variance:", np.var(marks, ddof=1))
```

## 5. Standard deviation — `np.std()`

**Definition:** Standard deviation is the square root of variance. It describes spread in the original units of the data.

**Analogy:** It tells us whether measurements tend to stay near their average or vary widely.

**Business example:** A delivery company can measure how consistently it meets delivery times.

**AI/ML use case:** Standard deviation is used when standardizing features.

The result is in the same unit as the original values (here, marks).

```python
marks = np.array([70, 80, 90])
print("Standard deviation:", np.std(marks))
```

## 6. Minimum and 7. Maximum — `np.min()` and `np.max()`

**Definition:** Minimum returns the smallest value; maximum returns the largest value.

**Analogy:** Finding the lowest and highest temperatures recorded in a week.

**Business example:** Find the smallest and largest daily sales.

**AI/ML use case:** Min-max scaling uses a feature's minimum and maximum to map values to a chosen range.

```python
temperatures = np.array([28, 31, 26, 35, 30])
print("Lowest temperature:", np.min(temperatures))
print("Highest temperature:", np.max(temperatures))
```

## 8. Argmin and 9. Argmax — `np.argmin()` and `np.argmax()`

**Definition:** `argmin` returns the index of the smallest value; `argmax` returns the index of the largest value. NumPy uses zero-based indexing.

**Analogy:** Instead of asking for the lowest score, ask which position contains the lowest score.

**Business example:** Find which day had the most orders.

**AI/ML use case:** `argmax` is often used to select the class with the highest predicted score.

```python
orders = np.array([15, 9, 22, 18])
print("Index of fewest orders:", np.argmin(orders))
print("Index of most orders:", np.argmax(orders))
print("Most orders:", orders[np.argmax(orders)])
```

## 10. Percentile — `np.percentile()`

**Definition:** A percentile gives a value at or below which a specified percentage of observations falls, according to NumPy's interpolation method.

**Analogy:** The 90th percentile of a test score shows a high score threshold that about 90% of scores are at or below.

**Business example:** A company may track the 95th percentile of page response times to understand slow experiences.

**AI/ML use case:** Percentiles can help detect outliers or choose clipping thresholds.

The 50th percentile is the median.

```python
response_times = np.array([100, 110, 120, 130, 300])
print("50th percentile:", np.percentile(response_times, 50))
print("90th percentile:", np.percentile(response_times, 90))
```

## 11. Dot product — `np.dot()`

**Definition:** For two one-dimensional arrays of equal length, the dot product multiplies matching values and adds the products.

**Formula:** `a[0]*b[0] + a[1]*b[1] + ...`

**Analogy:** Calculate a bill by multiplying each item's quantity by its price, then adding the costs.

**Business example:** Quantities multiplied by unit prices give total order value.

**AI/ML use case:** A linear model combines feature values with weights using a dot product.

```python
quantities = np.array([2, 3, 4])
unit_prices = np.array([100, 50, 25])
print("Total order value:", np.dot(quantities, unit_prices))
```

## 12. Matrix multiplication — `@` or `np.matmul()`

**Definition:** Matrix multiplication combines rows of the first matrix with columns of the second. The number of columns in the first matrix must equal the number of rows in the second.

**Analogy:** A table of quantities by product can be multiplied by a table of prices to calculate totals.

**Business example:** Calculate the total cost for several stores and products.

**AI/ML use case:** Neural networks use matrix multiplication to transform batches of input features with learned weights.

```python
quantities = np.array([[2, 3], [1, 4]])  # two stores, two products
prices = np.array([[10], [5]])           # price for each product
print("Store totals:")
print(quantities @ prices)
```

## 13. Transpose — `.T` or `np.transpose()`

**Definition:** Transpose swaps a matrix's rows and columns.

**Analogy:** Turning a table so that rows become columns and columns become rows.

**Business example:** Convert a table organized by stores and products into one organized by products and stores.

**AI/ML use case:** Transpose is used to align matrix dimensions for linear algebra calculations.

```python
sales_by_store = np.array([[10, 20, 30], [15, 25, 35]])
print("Original (stores x days):")
print(sales_by_store)
print("Transposed (days x stores):")
print(sales_by_store.T)
```

## 14. Basic linear algebra operations

Linear algebra works with vectors and matrices. NumPy's `numpy.linalg` module includes tools for solving systems, finding determinants and inverses, and computing vector lengths.

### Vector norm
**Definition:** A norm measures the size or length of a vector.

**Use case:** Vector lengths are useful when measuring distances or normalizing feature vectors.

```python
vector = np.array([3, 4])
print("Vector length (Euclidean norm):", np.linalg.norm(vector))
```

### Determinant
**Definition:** The determinant is a number calculated from a square matrix. It helps show whether a matrix is invertible: a nonzero determinant means it is invertible.

**Use case:** Determinants appear in linear systems and transformations.

```python
matrix = np.array([[2, 1], [1, 3]])
print("Determinant:", np.linalg.det(matrix))
```

### Solve a system of linear equations
**Definition:** `np.linalg.solve(A, b)` finds `x` such that `A @ x = b`, when the system has a unique solution.

**Business example:** Solve for unknown quantities when a small set of cost equations is known.

**AI/ML use case:** Linear algebra is a foundation for regression and many machine-learning algorithms.

```python
A = np.array([[2, 1], [1, 1]])
b = np.array([5, 3])
solution = np.linalg.solve(A, b)
print("Solution [x, y]:", solution)
print("Check A @ solution:", A @ solution)
```

### Matrix inverse
**Definition:** An inverse matrix `A_inv` satisfies `A @ A_inv = identity`, when the inverse exists.

**Note:** In practical code, solving a system with `np.linalg.solve()` is usually preferred to explicitly calculating an inverse.

```python
A = np.array([[2.0, 1.0], [1.0, 3.0]])
A_inverse = np.linalg.inv(A)
print("Inverse:")
print(A_inverse)
print("A multiplied by inverse (approximately identity):")
print(A @ A_inverse)
```
