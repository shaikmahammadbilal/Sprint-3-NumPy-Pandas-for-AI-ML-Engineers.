## Learning goals
Use NumPy's random number generator to create random values, select items, shuffle data, and sample from probability distributions.

**Recommended style:** Use `np.random.default_rng()` for new code. It creates a random number generator object.

```python
import numpy as np
rng = np.random.default_rng()
```

## 1. Random numbers — `rng.random()`

**Definition:** Generates random floating-point values from 0 (inclusive) up to 1 (exclusive).

**Real-world analogy:** Drawing a random point on a number line between 0 and 1.

**Simple example:** Generate one random decimal or a small array.

**Business example:** Simulate a random percentage or a basic uncertain quantity for a toy model.

**AI/ML use case:** Random values are used to initialize parameters or create random samples.

The output changes across runs unless the generator is initialized with a fixed seed.

```python
print("One random number:", rng.random())
print("Four random numbers:", rng.random(4))
```

## 2. Random seed — reproducible randomness

**Definition:** A seed initializes the random generator. The same seed and same sequence of operations produce the same sequence of random values.

**Analogy:** Starting a shuffled-card routine from the same reproducible starting state.

**Business example:** Reproduce a simulation result when reviewing a report.

**AI/ML use case:** Fixing a seed helps make experiments more reproducible, though some other libraries or hardware operations may still be nondeterministic.

```python
rng_a = np.random.default_rng(42)
rng_b = np.random.default_rng(42)

values_a = rng_a.random(5)
values_b = rng_b.random(5)

print("First generator:", values_a)
print("Second generator:", values_b)
print("Same values:", np.array_equal(values_a, values_b))
```

## 3. Random integers — `rng.integers()`

**Definition:** Generates random integers in a specified range. The lower bound is included and the upper bound is excluded.

**Analogy:** Rolling a fair six-sided die can be simulated by drawing integers from 1 through 6.

**Business example:** Randomly assign a sample customer to a test group for a simple demonstration.

**AI/ML use case:** Random integer indices can select records or mini-batches for experiments.

```python
die_rolls = rng.integers(1, 7, size=10)
print("Ten simulated die rolls:", die_rolls)
```

## 4. Random choice — `rng.choice()`

**Definition:** Selects one or more items from a given array or sequence.

**Analogy:** Picking a name from a bowl.

**Business example:** Select a small group of customers for a feedback survey.

**AI/ML use case:** Sample records for quick inspection or choose examples for a simple experiment.

By default, sampling is with replacement; set `replace=False` when items must not repeat and enough items are available.

```python
products = np.array(["Laptop", "Phone", "Tablet", "Watch"])
print("One selected product:", rng.choice(products))
print("Two selected products, no repeats:", rng.choice(products, size=2, replace=False))
```

## 5. Shuffle — `rng.shuffle()`

**Definition:** Randomly reorders an array in place, meaning the original array is changed.

**Analogy:** Shuffling a deck of cards.

**Business example:** Randomize the order in which survey responses are reviewed.

**AI/ML use case:** Shuffle training examples before dividing data into batches, when appropriate for the task.

```python
numbers = np.array([1, 2, 3, 4, 5])
print("Before shuffle:", numbers)
rng.shuffle(numbers)
print("After shuffle:", numbers)
```

## 6. Permutation — `rng.permutation()`

**Definition:** Returns a randomly permuted copy of an array, leaving the original array unchanged. For an integer `n`, it returns a permutation of integers from `0` to `n-1`.

**Analogy:** Create a shuffled copy of a list while keeping the original list intact.

**Business example:** Generate a randomized order for a product demonstration schedule.

**AI/ML use case:** Randomize row indices before creating a train/test split.

```python
original = np.array(["A", "B", "C", "D"])
shuffled_copy = rng.permutation(original)
print("Original:", original)
print("Permutation:", shuffled_copy)
print("Original unchanged:", original)
```

## 7. Random distributions

**Definition:** A probability distribution describes how likely different values are. NumPy can sample values from many distributions.

### Uniform distribution
Every value in a continuous interval has equal density.
**Example:** A simplified random waiting time between 0 and 10 minutes.

```python
uniform_samples = rng.uniform(low=0, high=10, size=5)
print("Uniform samples:", uniform_samples)
```

### Normal distribution
Values cluster around a mean, with spread controlled by the standard deviation. It is often drawn as a bell-shaped curve.

**Business example:** Simulate measurement noise around a target value. Real business data is not always normally distributed, so treat this as a model assumption.

**AI/ML use case:** Normal random values are commonly used for noise and some parameter initializations.

```python
normal_samples = rng.normal(loc=100, scale=15, size=8)
print("Normal samples:", normal_samples)
print("Sample mean:", np.mean(normal_samples))
```

### Binomial distribution
Counts the number of successes in a fixed number of independent trials, each with the same success probability.

**Business example:** Simulate how many customers out of 20 might click an offer if each has a 10% click probability.

**AI/ML use case:** Model binary outcomes such as success/failure counts.

```python
clicks = rng.binomial(n=20, p=0.10, size=10)
print("Simulated clicks across 10 groups of 20 customers:", clicks)
```

### Poisson distribution
Models counts of events in a fixed interval when events occur at an average rate under the distribution's assumptions.

**Business example:** Simulate the number of support tickets received in an hour when the average rate is 4 tickets per hour.

**AI/ML use case:** Count models can be useful for event data, when Poisson assumptions are reasonable.

```python
tickets = rng.poisson(lam=4, size=10)
print("Simulated hourly ticket counts:", tickets)
```

## 8. Combined example: simulate daily orders

This example models daily order counts with a Poisson distribution. The average rate (`lam=50`) is an assumption for this teaching example, not a forecast.

```python
rng_demo = np.random.default_rng(7)
daily_orders = rng_demo.poisson(lam=50, size=7)

print("Simulated orders for 7 days:", daily_orders)
print("Total orders:", np.sum(daily_orders))
print("Average orders per day:", np.mean(daily_orders))
```
