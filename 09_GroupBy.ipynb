# Notebook 9 – Grouping & Aggregation
**Filename:** `09_GroupBy.ipynb`

Grouping splits data into categories, calculates summaries for each category, and returns the results. This is useful for business reports and feature engineering.

```python
import pandas as pd

sales = pd.DataFrame({
    "Department": ["IT", "IT", "Sales", "Sales", "HR", "HR"],
    "Employee": ["Asha", "Ravi", "Mina", "John", "Zoya", "Arun"],
    "Salary": [55000, 60000, 35000, 38000, 32000, 34000],
    "SalesAmount": [0, 0, 12000, 18000, 0, 0]
})
display(sales)
```

## 1. `groupby()`
Groups rows with the same category. It is often followed by an aggregation such as `sum()`, `mean()`, or `count()`.

**Analogy:** Sort receipts into department folders, then total each folder.

```python
display(sales.groupby("Department")["Salary"].mean())
```

## 2. `agg()`
Applies one or more aggregation functions. Named aggregation gives readable output column names.

```python
summary = sales.groupby("Department").agg(
    AverageSalary=("Salary", "mean"),
    MaximumSalary=("Salary", "max"),
    EmployeeCount=("Employee", "count"),
    TotalSales=("SalesAmount", "sum")
)
display(summary)
```

## 3. `transform()`
Calculates a grouped result and returns values aligned to the original rows. Useful when adding group-level information to each record.

```python
sales["DepartmentAverageSalary"] = sales.groupby("Department")["Salary"].transform("mean")
display(sales)
```

## 4. `filter()`
Keeps or removes entire groups based on a condition evaluated for each group.

```python
# Keep departments with at least two employees
large_departments = sales.groupby("Department").filter(lambda group: len(group) >= 2)
display(large_departments)
```

## 5. Pivot tables
`pivot_table()` summarizes data with one variable as rows and another as columns. It can aggregate duplicate combinations.

```python
pivot = pd.pivot_table(
    sales,
    index="Department",
    values=["Salary", "SalesAmount"],
    aggfunc="mean",
    fill_value=0
)
display(pivot)
```

## 6. Crosstab
`pd.crosstab()` counts the frequency of combinations between categorical variables by default.

```python
staff = pd.DataFrame({
    "Department": ["IT", "IT", "Sales", "Sales", "HR", "IT"],
    "Level": ["Junior", "Senior", "Junior", "Junior", "Senior", "Junior"]
})
display(pd.crosstab(staff["Department"], staff["Level"]))
```

## Examples and interpretation
- **Business:** Average salary by department helps compare staffing costs.
- **AI/ML:** `transform()` can add a department-level average or group statistic as a feature for each employee record. Use care to avoid data leakage when features are used for prediction.
- **How the examples represent grouping:** Individual rows are combined into category-level summaries, or receive values calculated from their category.
