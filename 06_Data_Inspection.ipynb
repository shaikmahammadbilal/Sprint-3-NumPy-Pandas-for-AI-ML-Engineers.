# Notebook 6 – Data Inspection
**Filename:** `06_Data_Inspection.ipynb`

Data inspection means checking the size, structure, data types, missing values, and value distributions before analysis. It helps catch problems early.

```python
import pandas as pd

df = pd.DataFrame({
    "Employee": ["Asha", "Ravi", "Mina", "John", "Asha", "Ravi"],
    "Department": ["Sales", "IT", "IT", "HR", "Sales", "HR"],
    "Salary": [30000, 45000, 42000, 28000, 32000, 31000],
    "Experience": [1, 3, 2, 1, 2, 1]
})
display(df)
```

## `head()`
**Definition:** Shows the first rows (default five).

**Real-world meaning:** Check the first few sales records after importing a file.

```python
display(df.head())
```

## `tail()`
**Definition:** Shows the last rows (default five).

**Real-world meaning:** Check whether the file ends with expected records.

```python
display(df.tail(2))
```

## `sample()`
**Definition:** Returns random rows.

**Real-world meaning:** Spot-check records without always viewing the same rows.

```python
display(df.sample(2, random_state=1))
```

## `info()`
**Definition:** Shows row count, column names, non-null counts, and dtypes.

**Real-world meaning:** Find columns with missing values and unexpected types.

```python
df.info()
```

## `describe()`
**Definition:** Summarizes numeric columns by count, mean, standard deviation, min, quartiles, and max.

**Real-world meaning:** Review sales values and detect unusually large values.

```python
display(df.describe())
```

## `shape`
**Definition:** Returns (number_of_rows, number_of_columns).

**Real-world meaning:** Confirm the size of a dataset.

```python
display(df.shape)
```

## `columns`
**Definition:** Returns column labels.

**Real-world meaning:** Check imported field names.

```python
display(df.columns)
```

## `dtypes`
**Definition:** Returns each column's data type.

**Real-world meaning:** Find numeric values mistakenly imported as text.

```python
display(df.dtypes)
```

## `memory_usage()`
**Definition:** Estimates memory used by each column.

**Real-world meaning:** Understand memory needs for large datasets.

```python
display(df.memory_usage(deep=True))
```

## `nunique()`
**Definition:** Counts distinct values in each column by default.

**Real-world meaning:** Count unique customers or categories.

```python
display(df.nunique())
```

## `value_counts()`
**Definition:** Counts occurrences of each value in a Series.

**Real-world meaning:** Find the most common product category.

```python
display(df['Department'].value_counts())
```

## Recommended inspection workflow
1. `df.shape` – check dimensions.
2. `df.head()` – preview rows.
3. `df.info()` and `df.isna().sum()` – inspect types and missing values.
4. `df.describe()` – understand numeric ranges.
5. `value_counts()` and `nunique()` – inspect categories.

**AI/ML use case:** Before training a model, inspect data types, missing values, category frequencies, and numeric ranges. This represents data inspection because model quality depends on understanding the data before fitting the model.
