## 1. Create a NumPy array

**Question:** Create a NumPy array containing the numbers 10, 20, 30, 40, and 50.

```python
import numpy as np
import pandas as pd

numbers = np.array([10, 20, 30, 40, 50])
print(numbers)
```

**Answer:** `np.array()` converts the given sequence into a NumPy array. NumPy arrays are useful for efficient numerical calculations.

## 2. Difference between a NumPy array and a Python list

**Question:** Explain the difference and demonstrate it with code.

```python
python_list = [1, 2, 3]
numpy_array = np.array([1, 2, 3])

print("Python list:", python_list)
print("NumPy array:", numpy_array)
print("List repeated twice:", python_list * 2)
print("Array multiplied by 2:", numpy_array * 2)
```

**Answer:** A Python list is a general-purpose container and can hold different data types. A NumPy array is designed for numerical work and usually stores values of one data type. Multiplying a list by 2 repeats the list; multiplying an array by 2 multiplies each element.

## 3. Explain broadcasting with an example

**Question:** Add a single number to every value in a NumPy array.

```python
values = np.array([1, 2, 3, 4])
result = values + 10
print(result)
```

**Answer:** Broadcasting lets NumPy apply an operation between arrays of compatible shapes, including an array and a single number. Here, 10 is added to every element.

## 4. Difference between Series and DataFrame

**Question:** Create one Pandas Series and one DataFrame.

```python
scores = pd.Series([85, 90, 78], name="Score")
students = pd.DataFrame({
    "Name": ["Asha", "Ravi", "Meena"],
    "Score": [85, 90, 78]
})

print("Series:")
print(scores)
print("\nDataFrame:")
print(students)
```

**Answer:** A Series is one-dimensional (like a single column). A DataFrame is two-dimensional (like a table with rows and columns).

## 6. Merge two DataFrames

**Question:** Combine student details and scores using a common `StudentID` column.

```python
details = pd.DataFrame({
    "StudentID": [1, 2, 3],
    "Name": ["Asha", "Ravi", "Meena"]
})
marks = pd.DataFrame({
    "StudentID": [1, 2, 3],
    "Score": [85, 90, 78]
})

merged = pd.merge(details, marks, on="StudentID", how="inner")
print(merged)
```

**Answer:** `pd.merge()` joins DataFrames using matching key values. An inner merge keeps rows whose `StudentID` appears in both DataFrames.

## 7. Perform GroupBy and aggregation

**Question:** Find total sales for each department.

```python
sales = pd.DataFrame({
    "Department": ["Sales", "IT", "Sales", "IT", "HR"],
    "Amount": [1000, 1500, 700, 2000, 800]
})

total_by_department = sales.groupby("Department")["Amount"].sum()
print(total_by_department)
```

**Answer:** `groupby()` groups rows with the same department, and `sum()` adds the amounts in each group.

## 8. Filter rows using multiple conditions

**Question:** Select students whose score is at least 80 and whose attendance is at least 75.

```python
students = pd.DataFrame({
    "Name": ["Asha", "Ravi", "Meena", "John"],
    "Score": [85, 72, 90, 80],
    "Attendance": [90, 88, 70, 80]
})

filtered = students[(students["Score"] >= 80) & (students["Attendance"] >= 75)]
print(filtered)
```

**Answer:** Use `&` for AND between Pandas conditions, and put each condition inside parentheses. Both conditions must be true for a row to be selected.

## 9. Handle missing values

**Question:** Find missing values and replace missing numeric values with the column mean.

```python
data = pd.DataFrame({
    "Name": ["Asha", "Ravi", "Meena", "John"],
    "Age": [20, np.nan, 22, np.nan],
    "Score": [85, 90, np.nan, 75]
})

print("Missing values before filling:")
print(data.isna().sum())

data["Age"] = data["Age"].fillna(data["Age"].mean())
data["Score"] = data["Score"].fillna(data["Score"].mean())
print("\nData after filling missing numeric values with each column mean:")
print(data)
```

**Answer:** `isna().sum()` counts missing values in each column. `fillna()` replaces them. Filling with the mean is one option for numeric data; the best method depends on what the missing values mean.

## 10. Calculate statistical measures using NumPy

**Question:** Calculate mean, median, standard deviation, minimum, and maximum.

```python
marks = np.array([60, 70, 80, 90, 100])

print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Standard deviation:", np.std(marks))
print("Minimum:", np.min(marks))
print("Maximum:", np.max(marks))
```

**Answer:** These functions summarize a numeric dataset. `np.std()` uses population standard deviation by default (`ddof=0`). Use `np.std(marks, ddof=1)` for sample standard deviation.

## 11. Optimize a DataFrame for memory usage

**Question:** Check memory usage and reduce the memory used by suitable columns.

```python
data = pd.DataFrame({
    "Department": ["IT", "HR", "IT", "Sales", "HR", "IT"],
    "Quantity": [10, 20, 15, 10, 25, 30],
    "Price": [100.5, 200.0, 150.0, 100.5, 220.0, 175.5]
})

print("Memory usage before optimization (bytes):")
print(data.memory_usage(deep=True))
print("Total:", data.memory_usage(deep=True).sum())

# Category can save memory when a text column has repeated values.
data["Department"] = data["Department"].astype("category")
# Use a smaller integer type only when its range safely contains all values.
data["Quantity"] = pd.to_numeric(data["Quantity"], downcast="integer")

print("\nMemory usage after optimization (bytes):")
print(data.memory_usage(deep=True))
print("Total:", data.memory_usage(deep=True).sum())
print("\nData types:")
print(data.dtypes)
```

**Answer:** Check memory with `memory_usage(deep=True)`. Convert repeated text values to `category` when appropriate and downcast numeric columns only when the smaller type can safely hold all values. Memory savings depend on the size and structure of the real dataset.

## 15. Find duplicate rows

**Question:** Identify duplicate rows and remove exact duplicates.

```python
records = pd.DataFrame({
    "Name": ["Asha", "Ravi", "Asha", "Meena"],
    "Score": [85, 90, 85, 78]
})

print("Duplicate-row flags:")
print(records.duplicated())
clean_records = records.drop_duplicates()
print("\nAfter removing duplicates:")
print(clean_records)
```

**Answer:** `duplicated()` marks repeated rows (the first occurrence is not marked by default). `drop_duplicates()` returns a DataFrame with duplicate rows removed.

## 16. Practice challenge: Combine several operations

**Question:** Calculate average sales by department for rows where sales are at least 500.

```python
sales = pd.DataFrame({
    "Department": ["IT", "IT", "HR", "Sales", "HR", "Sales"],
    "Amount": [700, 300, 600, 900, 400, 500]
})

result = (sales[sales["Amount"] >= 500]
          .groupby("Department")["Amount"]
          .mean()
          .reset_index(name="AverageAmount"))
print(result)
```

**Answer:** First filter the rows, then group the remaining rows by department, calculate the mean, and use `reset_index()` to turn the result into a regular DataFrame.
