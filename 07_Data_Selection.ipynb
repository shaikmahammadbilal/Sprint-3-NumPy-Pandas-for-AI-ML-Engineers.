# Notebook 7 – Selecting & Filtering Data
**Filename:** `07_Data_Selection.ipynb`

Selecting chooses particular rows or columns. Filtering keeps rows that meet conditions.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Asha", "Ravi", "Mina", "John", "Zoya"],
    "Department": ["Sales", "IT", "IT", "HR", "Sales"],
    "Age": [24, 31, 27, 40, 29],
    "Salary": [30000, 55000, 48000, 42000, 36000]
}, index=[101, 102, 103, 104, 105])
display(df)
```

## 1. `loc`
Label-based selection: select rows and columns using their labels. The ending label in a label slice is included.

```python
display(df.loc[102])
display(df.loc[101:103, ["Name", "Salary"]])
```

## 2. `iloc`
Position-based selection: select rows and columns by integer position. The ending position in a slice is excluded.

```python
display(df.iloc[0])
display(df.iloc[0:3, [0, 3]])
```

## 3. Boolean indexing
Use a condition that produces `True` or `False` for each row. Only rows with `True` are kept.

```python
high_salary = df[df["Salary"] >= 45000]
display(high_salary)
```

## 4. Filtering rows
Combine conditions with `&` (AND), `|` (OR), and `~` (NOT). Put each condition in parentheses.

```python
display(df[(df["Age"] < 30) & (df["Salary"] >= 30000)])
display(df[(df["Department"] == "IT") | (df["Department"] == "Sales")])
```

## 5. `query()`
Filters rows using a string expression. `@variable` refers to a Python variable from the surrounding environment.

```python
minimum_salary = 40000
display(df.query("Salary >= @minimum_salary"))
```

## 6. `isin()`
Checks whether each value belongs to a list or set of accepted values.

```python
display(df[df["Department"].isin(["IT", "Sales"])])
```

## 7. `between()`
Checks whether values fall within a range. The endpoints are included by default.

```python
display(df[df["Age"].between(25, 35)])
```

## 8. Sorting data
`sort_values()` sorts by one or more columns. Use `ascending=False` for descending order.

```python
display(df.sort_values("Salary", ascending=False))
display(df.sort_values(["Department", "Age"]))
```

## Real-world examples
- **Analogy:** Filtering a spreadsheet to show only employees in IT is like applying a spreadsheet filter.
- **Business:** Find employees with salaries above a threshold or customers who purchased selected products.
- **AI/ML:** Select rows with valid target labels or filter records from a particular date range before model training.
- **How it represents the concept:** Each example keeps only the records relevant to a particular question or processing step.
