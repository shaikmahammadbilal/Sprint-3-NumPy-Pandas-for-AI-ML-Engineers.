## 1. What is NumPy & Why NumPy?

### **What is NumPy?**
NumPy stands for **Numerical Python**. It is a library used to store and process large sets of numbers arranged in tables called **Arrays**.

### **Why NumPy? (Why not use normal Python Lists?)**
1. **Super Fast Speed:** NumPy operations run up to **50 times faster** than simple Python lists.
2. **Continuous Memory Storage:** In normal lists, elements are scattered randomly in memory. NumPy blocks all numbers tightly side-by-side (contiguous memory), so your processor reads them instantly.
3. **Fixed Data Types:** A Python list can hold items of different types (numbers, text, etc.) which makes it heavy. NumPy arrays hold items of the **exact same type (homogeneous)**, which saves a massive amount of system memory.

### **Real-World Analogy:**
* **Python List:** Like a big messy bag where you throw clothes, shoes, and groceries together. Finding one specific thing takes time.
* **NumPy Array:** Like an organized egg tray where every single slot holds identical eggs lined up in perfect rows. You can count or pick them up instantly without any confusion.

```python
# First, we import the library. We give it a short name 'np' so it is easy to type.
import numpy as np
print("NumPy is loaded successfully!")
```

## 2. NumPy Arrays (`ndarray`) & Creating Arrays

The basic structure in NumPy is called an **`ndarray`** (N-dimensional array). You can easily convert a regular Python list into a NumPy array, or use built-in formulas to generate them from scratch.

### **Business Example:**
Let's track daily orders received by an online Biryani delivery store over 5 consecutive days.

```python
# 1. Making an array from a regular list
orders_list = [120, 145, 110, 160, 180]
orders_array = np.array(orders_list)
print("Daily Orders Array:", orders_array)

# 2. Making an array filled with Zeros (e.g., initial stock counter for 4 items)
initial_stock = np.zeros(4)
print("Initial Stock Array:", initial_stock)

# 3. Making an array with a continuous range of numbers (e.g., tokens 1 to 10)
tokens = np.arange(1, 11)
print("Token Numbers:", tokens)
```

---

## 3. Array Attributes: Dimensions, Shape, Size, and Data Types

To quickly check the properties of any array, we look at these four special attributes:
* **`ndim`**: Tells you the number of dimensions/axes (1D, 2D, etc.).
* **`shape`**: Shows the total rows and columns in a bracket format `(rows, columns)`.
* **`size`**: Tells you the total count of individual numbers inside.
* **`dtype`**: Tells you the data type of the entries (like integer or float).

### **AI/ML Use Case:**
In computer vision models, a simple grayscale photo is seen as a 2D array of rows and columns containing pixel brightness numbers. A colorful image (RGB) has 3 distinct layers, making it a 3D array.

```python
# Simulating a small 3x3 layout of pixel brightness values
pixels = np.array([
   [255, 0, 128],
   [200, 50, 0],
   [255, 255, 255]
])

print("Dimensions (ndim):", pixels.ndim)
print("Shape (rows, cols):", pixels.shape)
print("Total Elements (size):", pixels.size)
print("Data Type (dtype):", pixels.dtype)
```

## 4. Indexing and Slicing

* **Indexing:** Fetching a single specific value using its address position. Note: Counting always starts at **0**.
* **Slicing:** Cutting out a mini-portion of the array using `[start_index : end_index]`.

### **Simple Example:**
Let's extract information from weekly daytime temperature values.

```python
temps = np.array([32, 34, 31, 35, 36, 33, 30])

# Indexing: Get temperature of the absolute first day (index 0)
print("First Day Temperature:", temps[0])

# Slicing: Extract temperatures from Day 3 up to Day 5 (Indices 2 to 4)
print("Mid-week Temperatures:", temps[2:5])

# 2D Indexing: Fetching row 0, column 2 from our earlier pixel matrix
print("Specific Pixel Value:", pixels[0, 2])
```

---

## 5. Reshaping and Flattening

* **Reshaping (`.reshape()`):** Changing the grid arrangement (rows and columns) of data without modifying the actual numbers.
* **Flattening (`.flatten()`):** Crushing a complex multi-row grid structure back into a single long straight row.

### **AI/ML Use Case:**
When feeding pictures into standard Deep Learning networks, a 2D grid image (like 28x28 pixels) is always flattened into a single flat line of 784 numbers so that the network can read the features easily.

```python
# A straight 1D row with 6 elements
flat_prices = np.array([100, 200, 300, 400, 500, 600])

# Rearranging it into a 2D structure of 2 rows and 3 columns
grid_prices = flat_prices.reshape(2, 3)
print("Reshaped 2D Matrix:\n", grid_prices)

# Converting the 2D layout completely back to a basic line
back_to_flat = grid_prices.flatten()
print("Flattened row back again:", back_to_flat)
```

## 6. Copy vs View

This is a major logic point where bugs happen. Pay attention here:
* **View:** Just like looking at the exact same data from a different window. If you modify a view, the **original data changes too!** Normal slicing creates a View.
* **Copy:** Creates an entirely independent duplicate in memory. Modifying a copy leaves your **original data safe and untouched**.

### **Real-World Analogy:**
* **View:** Giving a friend permission to type inside your live Google Doc sheet. If they erase a line, it disappears for you too.
* **Copy:** Downloading that file as an offline Word Document and sending it to them. Whatever they edit on their own laptop will not corrupt your original file.

```python
original_data = np.array([10, 20, 30, 40])

# Let's check a View
mini_view = original_data[0:2]
mini_view[0] = 999
print("Original array changed because of View modification:", original_data)

# Let's reset the data and check a Copy
original_data = np.array([10, 20, 30, 40])
safe_copy = original_data.copy()
safe_copy[0] = 999
print("Original array stays perfectly safe when changing a Copy:", original_data)
```

## 7. Broadcasting

In simple math, you can only add or subtract arrays if they have the exact same shape. However, **Broadcasting** lets you perform mathematical calculations on arrays of completely different sizes. NumPy automatically expands or stretches the smaller value across the larger layout internally.

### **Business Example:**
Suppose you have the baseline menu prices of 3 food items, and you want to add a uniform festive festival service tax of ₹40 to each of them.

```python
menu_prices = np.array([250, 180, 350]) # 1D Array
tax = 40                                 # Single numeric value (scalar)

# Broadcasting automatically treats tax as [40, 40, 40] behind the scenes
final_bill = menu_prices + tax
print("Final billing amount after adding tax:", final_bill)
```

## 8. Array Operations & Universal Functions (`ufuncs`)

* **Vectorized Operations:** You can apply math calculations across entire arrays instantly, without writing slow, manual `for` loops.
* **Universal Functions (`ufuncs`):** Built-in math formulas in NumPy that work super fast on every single component of the array (e.g., `np.sqrt()`, `np.square()`, `np.log()`).

### **Simple Example:**
Let's instantly compute the geometric square root values for an array of plot measurements.

```python
land_areas = np.array([16, 25, 49, 100])

# Applying ufunc np.sqrt to compute side lengths all at once
sides = np.sqrt(land_areas)
print("Calculated side dimensions:", sides)

# Direct element-wise addition between two arrays
array_a = np.array([1, 2, 3])
array_b = np.array([10, 20, 30])
print("Sum of both arrays combined:", array_a + array_b)
```
