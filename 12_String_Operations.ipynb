# Notebook 12 – String Operations
## Learning objectives
Clean, search, split, replace, and validate text in Pandas using the `.str` accessor.

```python
import pandas as pd
customers = pd.DataFrame({
    "name": ["  ALICE Smith ", "bob JONES", "Chandra Rao  ", None],
    "email": ["alice@example.com", "bob@shop.in", "chandra@example.com", "unknown"],
    "city": ["Hyderabad", "Delhi", "Hyderabad", "Mumbai"]
})
print(customers)
```

## 1. `lower()`
**Definition:** Converts text to lowercase.
**Real-world:** Normalize names or email addresses before comparison.
**Business:** Avoid treating `SALES` and `Sales` as different categories.
**AI/ML:** Standardize text before text classification.

```python
customers["name_lower"] = customers["name"].str.lower()
print(customers[["name", "name_lower"]])
```

## 2. `upper()`
**Definition:** Converts text to uppercase.
**Real-world:** Standardize product codes.
**Business:** Make department labels consistent.
**AI/ML:** Normalize categorical text before feature preparation.

```python
customers["city_upper"] = customers["city"].str.upper()
print(customers[["city", "city_upper"]])
```

## 3. `strip()`
**Definition:** Removes whitespace from the beginning and end of each string.
**Real-world:** Clean user input with accidental spaces.
**Business:** Prevent duplicate customer categories caused by spaces.
**AI/ML:** Reduce inconsistent text values before encoding.

```python
customers["clean_name"] = customers["name"].str.strip()
print(customers[["name", "clean_name"]])
```

## 4. `split()`
**Definition:** Splits a string into pieces using a separator or pattern.
**Real-world:** Separate first and last names.
**Business:** Split a full product code into parts.
**AI/ML:** Extract simple tokens or fields from text.

```python
customers["email_domain"] = customers["email"].str.split("@").str[-1]
customers["first_name"] = customers["name"].str.strip().str.split().str[0]
print(customers[["email", "email_domain", "first_name"]])
```

## 5. `contains()`
**Definition:** Checks whether each string contains a specified substring or regular-expression pattern.
**Real-world:** Find messages mentioning a word.
**Business:** Identify company email addresses.
**AI/ML:** Create keyword-indicator features.

Missing strings produce missing results by default; `na=False` treats them as not matching.

```python
customers["is_example_email"] = customers["email"].str.contains("example", case=False, na=False)
print(customers[["email", "is_example_email"]])
```

## 6. `replace()`
**Definition:** Replaces matching text with new text.
**Real-world:** Fix a spelling variation.
**Business:** Standardize labels such as `Hyd.` and `Hyderabad`.
**AI/ML:** Normalize inconsistent text before modeling.

```python
customers["city_clean"] = customers["city"].str.replace("Hyd", "Hyderabad", regex=False)
print(customers[["city", "city_clean"]])
```

## 7. Regular Expressions (Regex)
**Definition:** A regular expression is a pattern used to find, validate, or replace text. Pandas string methods support regex patterns.

**Real-world:** Check whether a value resembles an email address.
**Business:** Extract order numbers from support messages.
**AI/ML:** Extract structured signals from text data.

This email pattern is a simple learning example, not a complete email-standard validator.

```python
emails = pd.Series(["user1@example.com", "bad-email", "sales.team@shop.in", None])
simple_email_pattern = r"^[^@\s]+@[^@\s]+\.[^@\s]+$"
print(emails.str.contains(simple_email_pattern, regex=True, na=False))
print("\\nDigits extracted:")
messages = pd.Series(["Order #1234 shipped", "Order #9821 pending", "No order number"])
print(messages.str.extract(r"#(\d+)"))
```
