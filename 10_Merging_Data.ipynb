# Notebook 10 – Combining Datasets
**Filename:** `10_Merging_Data.ipynb`

Combining datasets brings information from multiple tables together. The correct method depends on whether you want to stack rows or match records by keys.

```python
import pandas as pd

employees = pd.DataFrame({
    "EmployeeID": [1, 2, 3],
    "Name": ["Asha", "Ravi", "Mina"],
    "DepartmentID": [10, 20, 10]
})
departments = pd.DataFrame({
    "DepartmentID": [10, 20, 30],
    "Department": ["Sales", "IT", "HR"]
})
january = pd.DataFrame({"Product": ["Pen", "Book"], "Sales": [100, 250]})
february = pd.DataFrame({"Product": ["Pen", "Bag"], "Sales": [120, 300]})
display(employees)
display(departments)
```

## 1. `concat()`
Stacks DataFrames vertically (rows) or horizontally (columns). By default, it keeps the original index; `ignore_index=True` creates a fresh integer index.

**Analogy:** Put the contents of two monthly report pages into one longer report. This represents `concat()` because the rows are stacked rather than matched using a key.

```python
all_months = pd.concat([january, february], ignore_index=True)
display(all_months)
```

## 2. `merge()`
Combines tables by matching key columns, similar to a database SQL join. Specify `on`, `left_on`/`right_on`, and the join type (`how`).

- `inner`: only matching keys.
- `left`: every row from the left table, plus matches.
- `right`: every row from the right table, plus matches.
- `outer`: all keys from both tables.

```python
inner_result = pd.merge(employees, departments, on="DepartmentID", how="inner")
left_result = pd.merge(employees, departments, on="DepartmentID", how="left")
display(inner_result)
display(left_result)
```

## 3. `join()`
Joins DataFrames using their index by default. You can also use `on=` to match a left column to the right DataFrame's index.

```python
department_lookup = departments.set_index("DepartmentID")
joined = employees.join(department_lookup, on="DepartmentID")
display(joined)
```

## 4. `append()` – deprecation and alternatives
`DataFrame.append()` was deprecated in Pandas 1.4 and removed in Pandas 2.0. Do not use it in new code. Use `pd.concat([df1, df2], ignore_index=True)` instead.

```python
# Modern replacement for the old append pattern:
combined = pd.concat([january, february], ignore_index=True)
display(combined)
```

## 5. `combine_first()`
Fills missing values in one DataFrame with corresponding non-missing values from another, aligning by index and columns. Values already present in the first DataFrame are kept.

```python
old_profile = pd.DataFrame(
    {"Email": ["asha@example.com", None], "City": ["Ammanabrolu", None]},
    index=[1, 2]
)
new_profile = pd.DataFrame(
    {"Email": ["asha_new@example.com", "ravi@example.com"], "City": ["Hyderabad", "Guntur"]},
    index=[1, 2]
)
completed_profile = old_profile.combine_first(new_profile)
display(old_profile)
display(completed_profile)
```

## Real-world analogy
Think of an employee list and a department list:
- `concat()` adds more employee rows from another file.
- `merge()` matches each employee's department ID to the department name.
- `join()` is convenient when the lookup table is indexed by department ID.
- `combine_first()` fills gaps in an existing profile using a backup source.

## Simple example
The code above stacks January and February sales with `concat()`, then matches employee records with department names using `merge()`. This shows the difference between stacking rows and matching related records.

## Business example
A business can combine monthly sales reports, then merge product IDs with a product catalogue to display product names and categories. This helps create a unified report.

## AI/ML use case
A model-training table may combine customer activity with customer profile data by customer ID. Ensure keys are correct and check for duplicate keys, missing matches, and leakage from future information.

## Useful checks after combining
- Compare row counts before and after.
- Inspect missing values in merged columns.
- Check whether join keys are unique when expected.
- Use `indicator=True` in `merge()` to see where rows matched.

```python
merge_check = pd.merge(
    employees, departments, on="DepartmentID", how="outer", indicator=True
)
display(merge_check)
print(merge_check["_merge"].value_counts())
```
